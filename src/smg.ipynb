## Comparison of molecules


Running the comparison for all SMILES found in the workspace notebooks.

```python
all_smiles = [
    r"C1[C@H]([C@@H]1F)F",
    r"C1[C@H]([C@H]1F)F",
    # "CC[S@](=O)C",
    # "CC[S@@](=O)C",
]

for smi in all_smiles:
    print(f"\n ECFP of {smi}")
    svg_ecfp = visualize_molecules_with_atom_labels(
        [smi], radii=(0, 1, 2, 3), fingerprint_generator=ecfp
    )
    display(svg_ecfp)
    print(f"\n CircularStereoFingerprint of {smi}")
    svg = visualize_molecules_with_atom_labels(
        [smi], radii=(0, 1, 2, 3), fingerprint_generator=smg_fp
    )
    display(svg)
```
